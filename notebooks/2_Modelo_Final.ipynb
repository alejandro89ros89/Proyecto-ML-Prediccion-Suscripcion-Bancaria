## 1. Carga del modelo final

Cargamos el modelo entrenado y guardado en el Notebook 1 (`modelo_final.joblib`).
No es necesario volver a entrenar nada, ya que el pipeline incluye internamente el preprocesador (StandardScaler + OneHotEncoder + imputación).

```python
import joblib
import pandas as pd
import numpy as np

modelo_final = joblib.load('../modelos/modelo_final.joblib')
print("Modelo cargado correctamente.")
print(modelo_final)
```

## 2. Carga y preprocesado de los datos de competición

Cargamos el dataset ciego. Al igual que en el Notebook 1, aplicamos el mismo
tratamiento manual a la variable `pdays` antes de pasarla al pipeline:
creamos la variable binaria `contacto_previo` y sustituimos el -1 por 0.

```python
df_competicion = pd.read_pickle('../datos/bank_competition_29.pkl')

print(f"Dataset de competición cargado: {df_competicion.shape[0]} instancias, {df_competicion.shape[1]} variables.")

# Mismo preprocesado manual de pdays que en el Notebook 1
if (df_competicion['pdays'] == -1).any():
    df_competicion['contacto_previo'] = np.where(df_competicion['pdays'] == -1, 0, 1)
    df_competicion['pdays'] = np.where(df_competicion['pdays'] == -1, 0, df_competicion['pdays'])
    print("Variable 'pdays' preprocesada correctamente.")
```

## 3. Generación y guardado de predicciones

Utilizamos el modelo final para generar las predicciones sobre el conjunto
de competición y las guardamos en `predicciones.csv`.

```python
predicciones = modelo_final.predict(df_competicion)

df_predicciones = pd.DataFrame({
    'Id': df_competicion.index,
    'Prediction': predicciones
})

ruta_csv = '../predicciones.csv'
df_predicciones.to_csv(ruta_csv, index=False)

print(f"Predicciones guardadas en '{ruta_csv}'.")
print(f"Total de predicciones: {len(df_predicciones)}")
print(f"Distribución de clases:\n{df_predicciones['Prediction'].value_counts()}")
```
